## 🌐 Demo 5: Mini Servidor HTTP

Esta celda iniciará un servidor HTTP en Node.js que sirve una interfaz web para consultar Alchemy.

**⚠️ IMPORTANTE:**
- Actualiza `YOUR_API_KEY` en `demo-node-assets/.env` antes de ejecutar
- Esta celda bloqueará el kernel hasta que detengas el servidor (Ctrl+C o interrumpir kernel)
- Abre <http://localhost:8001> en tu navegador después de ejecutar

```python
from src.nodeRunner import NodeRunner

# Instalar dotenv para mini-server
runner_src = NodeRunner('./src')

# Instalar paquetes necesarios
print('📦 Instalando dependencias necesarias...')
runner_src.install_packages('dotenv')

# Ejecutar servidor (bloqueante)
print('🌐 Iniciando servidor Node.js...')
print('📍 Abre http://localhost:8001 en tu navegador')
print('⚠️  Presiona el botón ⏹️ (Stop) para detener el servidor\n')

!node src/mini-server.js
```
